### Using "Requests" to extract Raw Data from JSON file from GitHub. JSON file generated from Mockaroo


```python
# Importing "Requests" library

import requests
```

```python
# Assigning the raw file url a variable

patient_url = "https://raw.githubusercontent.com/Joseous/python_practice/refs/heads/main/mock_patient_data.json"
```

```python
content = requests.get(patient_url)
```

```python
# Checking if the "Requests" worked.

content.status_code
```

```python
# Printing the extracted Json data

patient_content = content.json()

patient_content
```

```python
# Checking the lenth of extracted Json data

len(patient_content)
```

```python
# Checking the type of extracted data

type(patient_content)
```

###  Converting the extracted JSON data into a well-structured DataFrame.

```python
# Import pandas library for data manipulation

import pandas as pd
```

```python
df = pd.DataFrame(patient_content)
```

```python
# Unsliced data

df
```

### Slicing the Data upto a 100 Rows

```python
sliced_patient_content = patient_content[:100]
```

```python
# Sliced extracted data

len(sliced_patient_content)
```

```python
df = pd.DataFrame(sliced_patient_content)
```

```python
df
```

```python
df.to_csv("patient_data.csv", index=False)
```

```python
import requests
import pandas as pd

response = requests.get("https://dummyjson.com/carts")
data = response.json()["carts"]

rows = []

for cart in data:
    for product in cart["products"]:
        product["cart_id"] = cart["id"]
        rows.append(product)

df = pd.DataFrame(rows)
```

```python
df
```

```python
df.to_csv("carts.csv", index=False)
```
